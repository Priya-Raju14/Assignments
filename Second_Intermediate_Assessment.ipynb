```python
# Import Libraries

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Loading Datasets

trainDF = pd.read_csv('train.csv')
testDF = pd.read_csv('test.csv')

print("Train Data shape : ", trainDF.shape)
print("Test Data shape : ", testDF.shape)
```

```python
# Exploratory Data Analysis

print("Train Data Info : \n")
print(trainDF.info())
print("\nTarget Value : \n")
print(trainDF['sexo'].value_counts())
```

```python
# Data Preprocessing
# Missing values

print("Train Data Missing Values : \n")
print(trainDF.isnull().sum())
print("\nTest Data Missing Values : \n")
print(testDF.isnull().sum())

# Filling missing values

trainDF=trainDF.fillna(trainDF.mean(numeric_only=True))
testDF=testDF.fillna(testDF.mean(numeric_only=True))

# After filling values

print("After - Train Data Missing Values : \n")
print(trainDF.isnull().sum())
print("\n After - Test Data Missing Values : \n")
print(testDF.isnull().sum())

# Separating features and target

x= trainDF.drop(['id','sexo'], axis=1)
y= trainDF['sexo']

```

```python
# Splitting train data

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# Train the model

model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(x_train, y_train)

# Evaluate

y_pred = model.predict(x_test)
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
print("Classification Report:\n", classification_report(y_test, y_pred))

# Predict

test_data = testDF.drop('id', axis=1)
predictions = model.predict(test_data)

# Download Final submission file

submission_df = pd.DataFrame({'id': testDF['id'], 'sexo': predictions})
submission_df.to_csv('submission.csv', index=False)

print("Submission File created Successfully \n")
print(submission_df.head())
print("Total rows in Submission file : ",len(submission_df))
```
