```python
# 1. Simple ATM Calculator
def calculator():
    balance = float(input("Balance: "))
    withdrawal = float(input("Withdrawal: "))

    if withdrawal <= balance:
        remaining_balance = balance - withdrawal
        print(f"Transaction successful. Remaining balance: ₹{remaining_balance:.0f}")
    else:
        print("Sorry,Insufficient balance.")
calculator()
```

```python
# 2. Electricity Bill Generator

def bill_generator():
    units = float(input("Enter units consumed: "))
    bill = 0

    if units <= 100:
        bill = units * 5
    elif units <= 200:
        bill = (100*5)+((units-100)*7)
    else:
        bill = (100*5)+(100*7)+((units-200)*10)

    print(f"Bill amount = ₹{bill:.0f}")

bill_generator()
```

```python
# 3. Temperature Converter

def converter():
    temp = float(input('Enter temperature value: '))
    unit = input('Enter unit (C/F): ')
    if unit == 'C':
        converted = (temp * 9 / 5) + 32
        print(f"Temperature in Fahrenheit: {converted:.1f}°F")
    elif unit == 'F':
        converted = (temp - 32) * 5 / 9
        print(f"Temperature in Celsius: {converted:.1f}°C")
    else:
        print("Invalid unit. Please enter 'C' or 'F'.")

converter()

```

```python
# 6. Sum of even numbers

def sum_of_even_numbers():
    num = int(input("Enter Number: "))
    total= 0

    for i in range(1, num + 1):
        if i % 2 == 0:  # Check if the number is even
            total += i

    print(f"Sum of even numbers: {total}")

sum_of_even_numbers()
```

```python
# 7. Dictionary – Student Marks Summary

def summary():
    student_marks = {}

    for i in range(3):
        name = input("Enter name: ")
        marks = float(input("Enter marks: "))
        student_marks[name] = marks

    total_marks = sum(student_marks.values())
    average = total_marks / len(student_marks)

    print(f"Average Marks: {average:.1f}")
summary()
```

```python
# 8. Dictionary – Word Frequency

def frequency():
    sentence = input("Sentence: ")

    words = sentence.split()
    frequency = {}
    for word in words:
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1

    for word, count in frequency.items():
        print(f"{word}: {count}")

# Execute function
frequency()
```

```python
# 9. Functions – Check Prime

def prime(number):
    if number <= 1:
        return False
    for i in range(2, number):
        if number % i == 0:
            return False
    return True

num = int(input("Enter a number : "))

if prime(num):
    print(f"{num} is a prime number.")
else:
    print(f"{num} is not a prime number.")
```

```python
# 10. Vowel Counter

def count(text):
    vowels = "aeiouAEIOU"
    count = 0

    for char in text:
        if char in vowels:
            count += 1
    return count

word = input("Enter a string: ")
vowels = count(word)
print(f"Number of vowels: {vowels}")
```

```python
# 11. Count Divisible Numbers

def divisibleNumbers():
    n = int(input("Enter the range limit (N): "))
    d = int(input("Enter the divisor (D): "))

    count = 0
    for i in range(1, n + 1):
        if i % d == 0:
            count += 1

    print(f"Count of numbers divisible by {d}: {count}")

divisibleNumbers()
```
